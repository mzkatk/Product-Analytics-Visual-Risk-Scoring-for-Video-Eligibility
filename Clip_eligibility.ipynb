```python
THRESHOLD = 65

df["model_decision"] = df.apply(
    lambda row: (
        "ELIGIBLE"
        if (row["duration_s"] <= 7) and (row["risk_score"] <= THRESHOLD)
        else "EXCLUDE"
    ),
    axis=1
)
```

```python
weights = {
    "mean": 0.40,
    "max_frame": 0.30,
    "max_hist": 0.30
}

def make_decision(risk):
    return (
        (df["duration_s"] <= 7) &
        (risk <= THRESHOLD)
    )

base_risk = 100 * (
    weights["mean"] * df["mean_frame_diff_pct"] +
    weights["max_frame"] * df["max_frame_diff_pct"] +
    weights["max_hist"] * df["max_hist_diff_pct"]
)

base_decision = make_decision(base_risk)

sensitivity_results = []

for name, value in weights.items():
    for direction, multiplier in [("-20%", 0.8), ("+20%", 1.2)]:

        modified = weights.copy()
        modified[name] = value * multiplier

        modified_risk = 100 * (
            modified["mean"] * df["mean_frame_diff_pct"] +
            modified["max_frame"] * df["max_frame_diff_pct"] +
            modified["max_hist"] * df["max_hist_diff_pct"]
        )

        modified_decision = make_decision(modified_risk)

        flips = (modified_decision != base_decision).sum()

        sensitivity_results.append({
            "constant": name,
            "direction": direction,
            "base_value": value,
            "modified_value": modified[name],
            "decision_flips": flips
        })

sensitivity_df = pd.DataFrame(sensitivity_results)

sensitivity_df
```

```python
def normalized_risk(w_mean, w_max_frame, w_max_hist):
    total = w_mean + w_max_frame + w_max_hist

    w_mean /= total
    w_max_frame /= total
    w_max_hist /= total

    return 100 * (
        w_mean * df["mean_frame_diff_pct"] +
        w_max_frame * df["max_frame_diff_pct"] +
        w_max_hist * df["max_hist_diff_pct"]
    )


sensitivity_results_norm = []

for name, value in weights.items():
    for direction, multiplier in [("-20%", 0.8), ("+20%", 1.2)]:

        modified = weights.copy()
        modified[name] = value * multiplier

        modified_risk = normalized_risk(
            modified["mean"],
            modified["max_frame"],
            modified["max_hist"]
        )

        modified_decision = (
            (df["duration_s"] <= 7) &
            (modified_risk <= THRESHOLD)
        )

        flips = (modified_decision != base_decision).sum()

        sensitivity_results_norm.append({
            "constant": name,
            "direction": direction,
            "base_value": value,
            "modified_value": modified[name],
            "decision_flips": flips
        })

pd.DataFrame(sensitivity_results_norm)
```
